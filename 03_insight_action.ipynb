# 핵심 인사이트 및 타겟 마케팅 전략
> 분석 결과를 바탕으로 **누구에게, 언제, 어떻게** 전화해야 하는지 실행 가능한 전략을 제안

## 비즈니스 문제 정의
은행 전화 마케팅 캠페인에서 무작위 전화는
비용만 높이고 성과는 낮습니다.
데이터 기반으로 가입 가능성이 높은
고객군을 정의하고 최적 전략을 도출합니다.

## KPI
| 지표 | 정의 | 수치 |
|------|------|------|
| 예금 가입률 | 전화 대상 중 실제 가입 비율 | 전체 평균 15.5% |
| 핵심 타겟 가입률 | 상위 타겟 조합 평균 | 60세 이상 42.3% |
| 비수기 성공률 | 5~8월 평균 | 11.7% |
| 최적 통화 성공률 | 11분 이상 통화 | 63.6% |


## 핵심 인사이트

### 1. 타겟 고객군
- 60세 이상 (주택대출X): 가입률 42.3% (평균 대비 2.7배)
- 학생: 35.4%
- 은퇴자 (대출X): 30.6%
→ 대출 부담 없는 고연령 + 학생이 핵심 타겟

### 2. 최적 마케팅 시기
- 3월, 9~12월 집중 운영
- 5~8월 최소화 (성공률 11.7%)

### 3. 통화 전략
- 통화 11분 이상 성공률 63.6%로 역전
- 연락 횟수 3회 이내가 가장 효율적
- 4회 이상부터 성공률 급감

## 서비스 액션 제안

### 1. 카테고리별 타겟 전략
- 60세 이상: 주택대출 여부 필수 확인 후 접근
- 학생: 전체 대상 캠페인 진행
- 29세 이하: 개인대출 없는 경우만 타겟

### 2. 시즌 전략
- 하반기(9~12월) 집중 운영
- 여름 시즌 캠페인 최소화

### 3. 통화 운영 전략
- 목표 통화 시간 11~27분
- 최대 연락 횟수 3회 이내 준수

## 한계점
- 실제 캠페인 비용 데이터 없어 ROI 계산 불가
- 인과관계 방향 불명확 (통화 길이 → 가입? or 관심 → 통화 길어짐?)
- 특정 시기 데이터 편중으로 월별 분석 신뢰도 제한
